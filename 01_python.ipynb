```python
def valeur_future (capital, taux, annees) :
    return capital*((1+taux))**annees;
    
print(valeur_future(1000, 0.05, 10));
```

```python
def valeur_actuelle(vf, taux, annees):
    return vf/((1+ taux))**annees;

print (valeur_actuelle(1628, 0.05, 10));
```

```python
def evolution_capital(vA, taux, annees):
    for i in range(1,annees+1):
        print(round(valeur_future(vA,taux,i),2))

evolution_capital(1000,0.03,20)

#autre méthode 

def evolution_capital2 (vAc, tau, annee):
    a=vAc;
    for i in range(1,annee+1):
        a=a*(1+tau)
        print("Année ",i,round(a,2))

evolution_capital2(1000,0.03,20)
```

```python
def moyenne(valeurs):
    return sum(valeurs)/len(valeurs)

print ("Moyenne ",moyenne([2, 4, 4, 4, 5, 5, 7, 9]))

def ecart_type(valeur):
    return (sum((v-moyenne(valeur) )**2 for v in valeur)/(len(valeur)-1))**0.5

print("Ecart type ", ecart_type([2, 4, 4, 4, 5, 5, 7, 9]))
```

```python
def frequence (lance):
    pile=0
    face=0
    for i in range(1,lance+1):
        a=random.randint(0,1)
        if a==1: #1 représente pile
            pile +=1
        else :
            face+=1

    freqPile=pile/lance
    freqFace=1-freqPile

    print("Fréquence pile : ", freqPile, "\n", "Fréquence face : ", freqFace)

frequence(100000000)
            

```

```python
def rendement (prix):
    resultat=[]
    for i in range(1,len(prix)):
        r=(prix[i]/prix[i-1])-1
        resultat.append(r)
    return resultat

print(rendement([100, 102, 99]))
```

```python
import csv

# 1. On initialise nos variables de mémorisation avant de fouiller le fichier
meilleur_prix = 0
meilleure_date = None

# 2. On ouvre le fichier (le bloc 'with' permet à Python de le refermer tout seul à la fin)
with open('prix.csv', newline='') as f:
    
    # 3. DictReader lit le fichier ligne par ligne comme un dictionnaire
    for ligne in csv.DictReader(f):
        
        # 4. CRUCIAL : On convertit le texte du CSV en nombre décimal (float)
        prix = float(ligne['prix'])
        
        # 5. On teste si le prix actuel est plus grand que notre record
        if prix > meilleur_prix:
            # Si oui, on met à jour nos variables de mémorisation
            meilleur_prix = prix
            meilleure_date = ligne['date']

# 6. On affiche le grand gagnant à la fin de la boucle
print(meilleure_date, meilleur_prix)
```

```python
def mensualité (montant, taux_annuel, annees):
    t=taux_annuel/12
    mois = annees*12
    return montant*(t/(1-(1+t)**-mois))

print(round(mensualité(200000, 0.035,20),2))
```
